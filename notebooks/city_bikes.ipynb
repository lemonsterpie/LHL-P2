# CityBikes

Send a request to CityBikes for the city of your choice. 

```python
import requests 
import os 
import pandas as pd
import json
from pprint import pprint
```

```python
url = 'https://api.citybik.es/v2/networks'
```

```python
request = requests.get(url)
data = request.json()
```

```python
network = data['networks']
```

```python
ids = []
names = []
latitudes = []
longitudes = []
cities = []
countries = []

for n in network:
    ids.append(n['id'])
    names.append(n['name'])
    latitudes.append(n['location']['latitude'])
    longitudes.append(n['location']['longitude'])
    cities.append(n['location']['city'])
    countries.append(n['location']['country'])

network_info = {'ids': ids,
                'names': names,
                'latitudes': latitudes,
                'longitudes': longitudes,
                'cities': cities,
                'countries': countries}

network_df = pd.DataFrame(network_info)
```
