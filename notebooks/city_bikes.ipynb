# CityBikes

Send a request to CityBikes for the city of your choice. 

```python
# Import packages 

import requests 
import os 
import pandas as pd
import json
```

```python
# Calling the CityBikes API for Vancouver and converting the results to json

vanurl = 'https://api.citybik.es/v2/networks/mobibikes'
vanrequest = requests.get(vanurl)
vandata = vanrequest.json()['network']['stations']
```

```python
# Example of a json entry

vandata[10]
```

Parse through the response to get the details you want for the bike stations in that city (latitude, longitude, number of bikes). 

```python
# The details I want are name, latitude, longitude, timestamp, free bikes, and empty slots

names = []
latitudes = []
longitudes = []
timestamps = []
bikes = []
spots = []

for item in vandata:
    names.append(item['name'])
    latitudes.append(item['latitude'])
    longitudes.append(item['longitude'])
    timestamps.append(item['timestamp'])
    bikes.append(item['free_bikes'])
    spots.append(item['empty_slots'])


# Example
print('Station Name: ', names[1])
print('Latitude: ', latitudes[1])
print('Longitude: ', longitudes[1])
print('Timestamp: ', timestamps[1])
print('Free bikes: ', bikes[1])
print('Empty Slots', spots[1])
```

Put your parsed results into a DataFrame.

```python
van_bikes = pd.DataFrame({'station_name': names,
                          'latitude': latitudes,
                          'longitude': longitudes,
                          'timestamp': timestamps,
                          'free_bikes': bikes,
                          'empty_slots': spots})
van_bikes
```
